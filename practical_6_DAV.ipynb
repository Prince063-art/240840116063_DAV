Aim : craete a basic data visualization (line,bar,histogram,scatter,pi)

step 1 : import libraries

```python
#import pandas
import pandas as pd
#import matplotlib
import matplotlib.pyplot as plt
# import numpy
import numpy as np
```

```python
data = {
    "Month":["January","February","March","April","May","June","July","August","September","October","November","December"],
    "Product":["Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Table","Laptop","Mobile","Table"],
    "Region":["North","North","North","South","South","South","East","East","East","West","West","West"],
    "Sales":[120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
    "Unit_sold":[12,30,25,14,35,30,18,38,32,21,45,40]
}
df=pd.DataFrame(data)
df
```

step 2 create a line chart
how did the company

```python
#company sales change from all months
plt.figure(figsize=(10, 5))
plt.plot(df["Month"], df["Sales"], marker="o", color="blue")
plt.title("Monthly Sales Trend (Jan to Dec)")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

step 3 create a barchart


which product generates how much total sales

```python
#total sales of product generates
product_sales = df.groupby("Product")["Sales"].sum()

plt.figure(figsize=(7, 5))
plt.bar(product_sales.index, product_sales.values, color=["blue", "red", "green"])
plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.show()
```

step 4 crerate a histogram

how are the individual sales value distributed

```python
#the individuals sales value
plt.figure(figsize=(7, 5))
plt.hist(df["Sales"], bins=5, color="yellow", edgecolor="black")
plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

step 5 create a scatterplot

is there a relation between the no of unit_sold and total sale

```python
#relation between the unit sold and sales
plt.figure(figsize=(7, 5))
plt.scatter(df["Unit_sold"], df["Sales"], color="red")
plt.title("Units Sold vs Sales")
plt.xlabel("Units Sold")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

step 6 create a pichart

what percentage of total sales come from each digit?

```python
#what percentage of total sales come from each digit?
region_sales = df.groupby("Region")["Sales"].sum()

plt.figure(figsize=(5, 6))
plt.pie(region_sales.values, labels=region_sales.index, autopct="%1.1f%%", startangle=80)
plt.title("Sales Share by Region")
plt.show()
```
