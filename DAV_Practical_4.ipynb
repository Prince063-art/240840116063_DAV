**Step 1: import libraries**

```python
#Import pandas
import pandas as pd
#Import matplotlib
import matplotlib.pyplot as plt
```

**Step 2: Create Dataset**

```python
#create dataset.
df={
    "Month": ["January", "Februay", "March", "April", "May", "June", "Jully", "August","September", "October", "NMovember", "December"],
    "Units_Sold": [120, 135, 150, 145, 160, 175, 180, 170, 190, 210, 230, 250],
    "Revenue": [120000, 135000, 150000, 145000, 160000, 175000, 180000, 170000, 190000, 210000, 230000, 250000],
    "Profit": [20000, 23000, 52000, 24000, 27000, 30000, 31000, 29000, 33000, 37000, 41000, 45000]
}

df = pd.DataFrame(df)
df
```

**Step 3: Inspect The Dataset**

```python
#Display First Five rows.
df.head()
```

```python
# no of rows and columns.
df.shape
```

```python
#Display name  columns.
df.columns
```

```python
#Display information.
df.info()
```

**Step 4: Generate Discriptive Statistic**

```python
#Statistics of Numeric Columns.
df.describe()
```

**Step 5:  Claculate the avg no. of units sold per month**

```python
# Avg no of Units sold per month.
average_units = df["Units_Sold"].mean()
print("Average no. of units sold per month:", average_units)
```

**Step 6: Calculate Median**

```python
#Calculate the median.
median_units = df["Units_Sold"].median()
print("Median no. of units sold per month:", median_units)
```

**Step 7: Calculate Max Min**

```python
#Find thye highest units sales in month.
maximun_units = df["Units_Sold"].max()

#no. of units sold in mounth.
minimun_units = df["Units_Sold"].min()

print("Maximum no. of units sold in a month:", maximun_units)
print("Minimum no. of units sold in a month:", minimun_units)
```

**Step 8: Calculate Standard Deviation**

```python
#Display Standard Daviation and calculate
std_unit = df["Units_Sold"].std()
print("Standard Deviation:", std_unit)
```

**Step 9: Find thye mounth with Highest Sales**

```python
#Find the index of row having max sales.
max_revenue_index = df["Revenue"].idxmax()
print("Row index with highest sales:", max_revenue_index)
month_with_highest_sales = df.loc[max_revenue_index, "Month"]
highest_revenue = df.loc[max_revenue_index, "Revenue"]

print("Month with highest sales:", month_with_highest_sales)
print("Highest revenue:", highest_revenue)
```

**Step 10: Find the Month with lowest sale**

```python
#Find the index of row having min revenue.
min_revenue_index = df["Revenue"].idxmin()
#Display the result.
print("Row index with lowest sales:", min_revenue_index)

#Retrieve the month corresponding
month_with_lowest_sales = df.loc[min_revenue_index, "Month"]

lowest_revenue = df.loc[min_revenue_index, "Revenue"]

print("Month with lowest sales:", month_with_lowest_sales)
print("Lowest revenue:", lowest_revenue)
```

**Step 11: Calculate total Revenue**

```python
#Diaplay total Revenue od this
total_revenue = df["Revenue"].sum()
print("Total Revenue:", total_revenue)
```

**Step 12: Calculate the total Profit Generated per year**

```python
#Display Total Profit for all years
total_profit = df["Profit"].sum()
print("Total Profit:", total_profit)
```

**Step 13: Identify the Sales Trend**

```python
#create a Line Chat to identify the sales trend.
plt.figure(figsize=(10, 5))

plt.plot(df["Month"], df["Revenue"], marker="o")

plt.title("Monthly Revenue Trend")

plt.xlabel("Month")

plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.show()
```

**Step 14: Analyze the Profit Trend**

```python
#Create a Line chart for Month Profit detail.
plt.figure(figsize=(10, 5))

plt.plot(df["Month"], df["Profit"], marker="o")

plt.title("Monthly Profit Trend")

plt.xlabel("Month")

plt.ylabel("Profit")

plt.xticks(rotation=45)

plt.show()
```

**Step 15: Compare Revenue And Profit**

```python
#Create a Line chart to comparision between profit and revanue.
plt.figure(figsize=(10, 5))

plt.plot(df["Month"], df["Revenue"], marker="o", label="Revenue")

plt.plot(df["Month"], df["Profit"], marker="o", label="Profit")

plt.title("Comparison of Revenue and Profit")

plt.xlabel("Month")

plt.ylabel("Amount")

plt.legend()

plt.show()
```

**Step 16:Identify Growth in Sales**

```python
#Calculate the percentage change in  Revenue
df["Revenue_Growth_%"] = df["Revenue"].pct_change() * 100
df
```

**Step 17: Identify the Month With Increasing Revenue**

```python
#Filter Dataset to display increasing revanue.
increasing_sales = df[df["Revenue_Growth_%"]>0]
increasing_sales
```
