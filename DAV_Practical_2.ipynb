step - 1 : import Pandas Library

```python
# import pandas
import pandas as pd
```

step-2:Create dataframe

```python
#
data = {
    "Id": [101,102,103,104,104,105,106,107],
    "Name": ["Rahul", "Priya", "Amit", None, "Riya","Ajay","Naresh","Jenil"],
    "Age": [20,21,None,20,20,22,23,28],
    "City": ["Surat","Navsari","Surat","Bharuch","Bharuch","Navsari","Surat","Ahemdabad"],
    "Marks": [85,None,78,88,88,81,90,93]
}

df = pd.DataFrame(data)
df
```

step 4:check missing values

```python
#Check for mission values
df.isnull()
```

```python
#print null values for each column
df.isnull().sum()
```

```python
#print the total null values in dataset
df.isnull().sum().sum()
```

step 5:fill missing values

```python
#Replace missing values in marks column.
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
#Display the updated dataframe
df
```

```python
#Replace missing values in the age column.
df["Age"] = df["Age"].fillna(df["Age"].mean())
#Display the updated dataframe
df
```

```python
#Replace missing names with the text "unknown".
df["Name"] = df["Name"].fillna("Unknown")
#Display the updated dataframe
df
```

step 6:Detect duplicate rercords

```python
#detect the duplicate values
df.duplicated()
```

```python
#count the total number of duplicate rows
df.duplicated().sum()
```

step 7: remove the dupolicate records

```python
#Drop the duplicate values
df.drop_duplicates()
df
```

step 8:rename the column.

```python
#rename the column "Marks" to "Score"
df.rename(columns={"Marks":"Score"},inplace=True)
df
```

step 9:Change data type

```python
#Convert the age column from float to integer
df["Age"] = df["Age"].astype(int)
df
```

step 10:Create a new column

```python
#Create a new column.
df["Final_Score"] = df["Score"] + 5
df
```

Step - 11 Convert text in uppercase

```python
# convert into uppercase
df["Name"] = df["Name"].str.upper()
df
```

Step - 12 Convert text into lower case

```python
# convert into lowercase
df["City"] = df["City"].str.lower()
df
```

Step - 13 Remove extra space

```python
#removes the extra space.
df["City"] = df["City"].str.strip()
df
```

Step - 14 sort the dataset

```python
# sort the dataset values
df.sort_values(by=['Final_Score'], inplace=True)
df
```

Step - 15 Save the clean data




```python
# Save Clean DataFrame
df.to_csv('Cleaned_Students.csv', index=False)
df
```
