**Aim : Create interactive visualizations using Plotly Express and Plotly Graph Objects.**

**step - 1 import the requred libraries**


```python
#import pandas library
import pandas as pd
#import plotly express library
import plotly.express as px
#import plotly graph library
import plotly.graph_objects as go
```

**step -2 create the dataset**

```python
#create dataset
data = {
    "Month": ["januray", "february", "march", "april", "may", "june", "july", "august", "september", "october", "november", "december"],
    "Product": ["Laptop", "Tablet", "mobile", "Laptop", "Tablet", "mobile", "Laptop", "Tablet", "mobile", "Laptop", "Tablet", "mobile"],
    "Region": ["north", "south", "east", "west","north", "south", "east", "west","north", "south", "east", "west"],
    "sales": [120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
    "Units_Sold": [12, 30, 25, 14, 35, 30, 18, 38, 32, 21, 45, 40],
    "Profit": [18000, 27000, 30000, 22000, 32000, 35000, 28000, 34000, 38000, 33000, 46000, 50000]
}
df = pd.DataFrame(data)
df
```

**Step 3: Inspect the Dataset**

```python
#display first five records from the dataset
df.head()
```

```python
#display information about the dataset
df.info()
```

```python
#display rows and columns
df.shape
```

**Step - 4 create an interective line chart using plotly express**

```python
fig = px.line(
    df,
    x='Month',
    y='sales',
    markers=True,
    title='Monthly Sales Revenue'
)

fig.update_layout(
    xaxis_title='Month',
    yaxis_title='Sales Revenue',
    hovermode='x unified'
)

fig.show()
```

**step - 5 improve the hover information**

```python
import plotly.express as px
fig = px.line(df, x="Month", y="sales", markers=True, title="Monthly Sales Trend with Extra Hover Details", hover_data=["Profit", "Units_Sold", "Region"])
fig.show()
```

**Step 6: Create an Interective Bar Chart**

```python
# Group the dataset by Product and calculate total Sales
product_sales = df.groupby("Product", as_index=False)["sales"].sum()

# Create interactive bar chart
fig = px.bar(
    product_sales,
    x="Product",
    y="sales",
    title="Total Sales Generated by Each Product",
    text="sales",
    hover_data=["sales"]
)

fig.update_layout(
    xaxis_title="Product",
    yaxis_title="Total Sales",
    hovermode="x"
)

fig.show()
```

**Step 7: Interective Scatter Plot**

```python
fig = px.scatter(
    df,
    x="Units_Sold",
    y="sales",
    title="Relationship Between Units Sold and Sales",
    hover_data=["Month", "Product", "Region", "Profit"],
    size="Profit"
)

fig.update_layout(
    xaxis_title="Units Sold",
    yaxis_title="Sales",
    hovermode="closest"
)

fig.show()
```

**step 8 interactive scatter plot with color**

```python
#create interactive scatter plot of unit_sold and sales with colors.
fig = px.scatter(
    df,
    x="Units_Sold",
    y="sales",
    color="Region",
    title="Relationship Between Units Sold and Sales",
    hover_data=["Month", "Product", "Region", "Profit"],
    size="Profit",

    )
fig.show()
```

**step 9 create an interective pie chart**


```python
# Calculate total sales of region
region_sales = df.groupby("Region", as_index=False)["sales"].sum()

# Create pie chart
fig = px.pie(
    region_sales,
    names="Region",
    values="sales",
    title="Regional Sales Contribution",
    hover_data=["sales"],
    hole=0
)
fig.show()
```
