Step - 1 Import Pandas library

```python
# import pandas
import pandas as pd
```

Step - 2 Create a dataset

```python
# Create a dataset
data = {
    "Id": [101,102,103,104,104,105,106,107],
    "Name": ["Rahul", "Priya", "Amit", None, "Riya","Ajay","Naresh","Jenil"],
    "Age": [20,21,None,20,20,22,23,28],
    "City": ["Surat","Navsari","Surat","Bharuch","Bharuch","Navsari","Surat","Ahemdabad"],
    "Marks": [85,None,78,88,88,81,90,93]
}
```

Step - 3 Create and display dataset

```python
# Create a data frame using given data.
df = pd.DataFrame(data)
df
```

Step - 4 Save dataset as csv file

```python
# Save and display the dataset in csv file.
df.to_csv("student.csv",index=False)

print("CSV file create successfully")
```

Step - 5 Save Dataframe As Excel

```python
#Save and display the dataset into Excel.
df.to_excel("student.xlsx",index=False)
#Display
print("Excel file create successfully")
```

Step - 6 Save dataframe As Json

```python
#Save and display the dataset into Json.
df.to_json("student.json",orient="records", indent=4)
#Display
print("JSON file create successfully")
```
